```python
# задание номер 6
def parse_price(text):
    text = text.strip().replace(' ','').replace(',','.')
    try:
        return float(text)
    except ValueError:
        return None



for sample in ["320.00", "320,00", " 320 ", "1 250,50", "n/a", ""]:
    print(f"{sample!r:>12} -> {parse_price(sample)}")
```

```python
import random
from pathlib import Path

random.seed(7)

menu = [
    ("Латте", 320), ("Капучино", 300), ("Американо", 220), ("Раф", 380),
    ("Эспрессо", 180), ("Круассан", 240), ("Чизкейк", 420), ("Сэндвич", 390),
    ("Какао", 280), ("Чай", 200),
]

lines = ["order_id;date;item;qty;price"]

for i in range(1, 241):
    order_id = f"{i:04d}"
    date = f"2025-03-{random.randint(1, 31):02d}"
    name, price = random.choice(menu)
    qty = random.choices([1, 2, 3, 4], weights=[55, 28, 12, 5])[0]

    # название товара: где-то лишние пробелы, где-то другой регистр
    r = random.random()
    if r < 0.12:
        item_str = "  " + name.lower() + " "
    elif r < 0.18:
        item_str = name.upper()
    else:
        item_str = name

    qty_str = str(qty)
    price_str = f"{price}.00"

    # дефекты выгрузки
    d = random.random()
    if d < 0.04:
        qty_str = ""                                   # количество не выгрузилось
    elif d < 0.07:
        qty_str = random.choice(["две", "n/a", "-"])    # количество текстом
    elif d < 0.13:
        price_str = f"{price},00"                       # запятая вместо точки
    elif d < 0.16:
        price_str = "n/a"                               # цена не выгрузилась
    elif d < 0.19:
        qty_str = "-1"                                  # возврат записан как минус
    elif d < 0.22:
        lines.append(f"{order_id};{date};{item_str};{qty_str}")   # потеряна колонка
        continue
    elif d < 0.25:
        qty_str = f" {qty} "                            # лишние пробелы

    lines.append(f"{order_id};{date};{item_str};{qty_str};{price_str}")

Path("data").mkdir(exist_ok=True)
Path("data/sales_raw.csv").write_text("\n".join(lines), encoding="utf-8")

print("Готово. Создан файл data/sales_raw.csv")
print("Строк вместе с заголовком:", len(lines))
print()
print("Первые 8 строк файла:")
for line in lines[:8]:
    print("   ", line)
```

```python
text = open("data/sales_raw.csv", encoding="utf-8").read()
lines = text.split("\n")

header = lines[0]
rows = lines[1:]

print("заголовок:", header)
print("строк с данными:", len(rows))
print()
print("первые 5 строк:")
for line in rows[:5]:
    print("   ", line)
```

```python
# задание номер 8
good = 0
bad_structure = 0
bad_number = 0
bad_logic = 0

revenue = 0.0
revenue_by_item = {}

for line in rows:
   
    parts = line.split(";")

    if len(parts)!=5:
        bad_structure+=1
        continue

    item = parts[2].strip().lower()

    try:
        qty = float(parts[3])
        price = parse_price(parts[4])

        if price is None:
            bad_number += 1
            continue
        
    except ValueError:
        bad_number+=1
        continue

    if qty<=0 or price<=0:
        bad_logic+=1
        continue

    cost = qty*price

    revenue +=cost
    good +=1
    revenue_by_item[item] = revenue_by_item.get(item,0)+cost

print("успешно разобрано:", good)
print("битая структура:  ", bad_structure)
print("не число:         ", bad_number)
print("нелогичные данные:", bad_logic)
print(f"выручка: {revenue:.2f}")
```

```python
total_rows = len(rows)
bad_total = bad_structure + bad_number + bad_logic

average_check = revenue / good
defective_rate= (bad_total / total_rows)*100

top_3 = sorted(revenue_by_item.items(),key=lambda pair: pair[1], reverse=True)
print("ОТЧЁТ ПО ПРОДАЖАМ ЗА МАРТ 2025")
print("-" * 40)
print(f'Строк в файле - {total_rows}')
print(f'разобрано - {good}')
print(f"выручка: {revenue:.2f}")
print(f'Средний чек - {average_check:.2f}')
print(f'Доля брака - {defective_rate:.1f}%')
print()
print('ТОП 3 ТОВАРА')
print('-'* 40)
for i, (item_top3, revenue_top3) in enumerate(top_3[:3],1):
    print(f'{i}.{item_top3} - {revenue_top3:.2f}')
```
